```python
# check outliers

num_col = list(data.select_dtypes("number").columns)
result = {}

for column in num_col : 
    
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    
    IQR = Q3 - Q1 
    
    lower = Q1 - (IQR * 1.25)
    upper = Q3 + (IQR * 1.25)
    
    outliers = data[(data[column] > upper) | (data[column] < lower)]
    
    result[column] = result.get(column , {
        "lower" : round(float(lower) , 2), 
        "upper" : round(float(upper) , 2), 
        "outliers_num" : len(outliers)
        }) 
    

result
```

```python
# check outliers values : 

for key , value in result.items() :
    
    upper =  value["upper"]
    lower = value["lower"]
    
    print(f" {key} : {data[(data[key] > upper) | (data[key] < lower)][key].values} ", end="\n\n\n")
```
