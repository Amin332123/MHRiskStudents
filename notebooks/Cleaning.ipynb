```python
# replace screen time hours  with social media hours (only if first > second ) : 


def replace_screentime_with_socialMediaHours() :
     # check social media hours with screen time hours :(37 rows)
     data.loc[data["Social_Media_Hours"] > data["Screen_Time_Hours"] , ["Social_Media_Hours", "Screen_Time_Hours"]]



     data.loc[data["Social_Media_Hours"] > data["Screen_Time_Hours"], "Screen_Time_Hours"] = data["Social_Media_Hours"]

replace_screentime_with_socialMediaHours()
# now social media hours = screen time hours . 
data.iloc[78] 

```
